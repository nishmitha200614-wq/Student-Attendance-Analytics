```python
"""Q4. Convert Attendance % from string format (e.g., "64%") to a numeric float column (0.64 or 64.0). Also convert Last Present Date to datetime, handling "Never" as NaT."""
#-----------------------------Write Solution Code-----------------------------

df = pd.read_excel("CSEB_Attendance.xlsx")
df["Attendance %"] = df["Attendance %"].astype(str).str.replace("%", "").astype(float) / 100
df["Last Present Date"] = pd.to_datetime(
    df["Last Present Date"].replace("Never", pd.NaT),
    errors="coerce"
)
print(df[["Attendance %", "Last Present Date"]].head(10))
print("\nData types:")
print(df[["Attendance %", "Last Present Date"]].dtypes)
```

### Section 3: Grouping & Aggregation 

```python
"""Q9. Group the data by Status and compute: 
a) Count of students 
b) Average attendance %  
c) Maximum consecutive absences"""
#-----------------------------Write Solution Code-----------------------------

df = pd.read_excel("CSEB_Attendance.xlsx", sheet_name="Student Summary")
df["Attendance %"] = (
    df["Attendance %"].astype(str).str.replace("%", "").astype(float)
)
result = df.groupby("Status").agg(
    Student_Count=("Student Name", "count"),
    Average_Attendance=("Attendance %", "mean"),
    Max_Consecutive_Absences=("Consecutive Absences", "max")
)
print(result)
```

```python
"""Q10. Create a new column Attendance Category:
≥ 90% → "Excellent"
75-89% → "Good"
60-74% → "Average"
< 60% → "Poor"
Then use groupby() to count students in each category."""
#-----------------------------Write Solution Code-----------------------------

df = pd.read_excel("CSEB_Attendance.xlsx", sheet_name="Student Summary")
df["Attendance %"] = (
    df["Attendance %"].astype(str).str.replace("%", "").astype(float)
)
def category(attendance):
    if attendance >= 90:
        return "Excellent"
    elif attendance >= 75:
        return "Good"
    elif attendance >= 60:
        return "Average"
    else:
        return "Poor"

df["Attendance Category"] = df["Attendance %"].apply(category)
result = df.groupby("Attendance Category")["Student Name"].count()
print("Number of students in each category:")
print(result)
```

```python
"""Q11. Calculate the total number of classes conducted (sum of Total Classes) and the total present/absent counts across all students using .agg()."""
#-----------------------------Write Solution Code-----------------------------

df = pd.read_excel("CSEB_Attendance.xlsx", sheet_name="Student Summary")
result = df.agg({
    "Total Classes": "sum",
    "Present": "sum",
    "Absent": "sum"
})
print("Total Classes, Present, and Absent:")
print(result)
```

```python
"""Q12. Group by Status and use .agg() to find:
a)min, max, and mean of Present
b)sum of Absent"""
#-----------------------------Write Solution Code-----------------------------

df = pd.read_excel("CSEB_Attendance.xlsx", sheet_name="Student Summary")
result = df.groupby("Status").agg({
    "Present": ["min", "max", "mean"],
    "Absent": "sum"
})
print(result)
```

### Section 4: Feature Engineering & Transformation

```python
"""Q13. Create a new column Risk Score using the formula:
Risk Score = (Absent * 0.4) + (Consecutive Absences * 0.6)
Display the top 10 students with the highest risk score."""
#-----------------------------Write Solution Code-----------------------------

df = pd.read_excel("CSEB_Attendance.xlsx", sheet_name="Student Summary")
df["Risk Score"] = (df["Absent"] * 0.4) + (df["Consecutive Absences"] * 0.6)
top_10 = df.sort_values(by="Risk Score", ascending=False).head(10)
print("Top 10 Students with Highest Risk Score:")
print(top_10[["Roll No", "Student Name", "Absent", "Consecutive Absences", "Risk Score"]])
```

```python
"""Q14. Add a column Classes Needed (Numeric) by extracting the numeric value from the Classes Needed (75%) column (e.g., "5 more" → 5, "✓ Met" → 0)."""
#-----------------------------Write Solution Code-----------------------------

df = pd.read_excel("CSEB_Attendance.xlsx", sheet_name="Student Summary")
df["Classes Needed (Numeric)"] = (
    df["Classes Needed (75%)"]
    .str.extract(r"(\d+)", expand=False)
    .fillna(0)
    .astype(int)
)
print(df[["Student Name", "Classes Needed (75%)", "Classes Needed (Numeric)"]])
```

```python
"""Q15. Using apply() with a lambda function, create a column Attendance Flag that returns:
a)"Defaulter" if attendance < 75%
b)"Safe" otherwise """
#-----------------------------Write Solution Code-----------------------------

df = pd.read_excel("CSEB_Attendance.xlsx", sheet_name="Student Summary")
df["Attendance %"] = (
    df["Attendance %"].astype(str).str.replace("%", "").astype(float)
)
df["Attendance Flag"] = df["Attendance %"].apply(
    lambda x: "Defaulter" if x < 75 else "Safe"
)
print(df[["Student Name", "Attendance %", "Attendance Flag"]])
```

```python
"""Q16. Use pd.cut() to bin students into attendance ranges: [0-50, 50-75, 75-90, 90-100] and label them as ["Critical", "Warning", "Good", "Excellent"]."""
#-----------------------------Write Solution Code-----------------------------

df = pd.read_excel("CSEB_Attendance.xlsx", sheet_name="Student Summary")
df["Attendance %"] = (
    df["Attendance %"].astype(str).str.replace("%", "").astype(float)
)
bins = [0, 50, 75, 90, 100]
labels = ["Critical", "Warning", "Good", "Excellent"]

df["Attendance Range"] = pd.cut(
    df["Attendance %"],
    bins=bins,
    labels=labels,
    include_lowest=True
)
print(df[["Student Name", "Attendance %", "Attendance Range"]])
```

### Section 5: Advanced Analysis 

```python
"""Q17. Using pivot_table(), create a summary showing the count of students by Status and Attendance Category (created in Q10)."""
#-----------------------------Write Solution Code-----------------------------

df = pd.read_excel("CSEB_Attendance.xlsx", sheet_name="Student Summary")
df["Attendance %"] = (
    df["Attendance %"].astype(str).str.replace("%", "").astype(float)
)
df["Attendance Category"] = df["Attendance %"].apply(
    lambda x: "Excellent" if x >= 90 else
              "Good" if x >= 75 else
              "Average" if x >= 60 else
              "Poor"
)
result = pd.pivot_table(
    df,
    index="Status",
    columns="Attendance Category",
    values="Student Name",
    aggfunc="count",
    fill_value=0
)
print(result)
```

```python
"""Q18. Identify duplicate student names (e.g., "ANANYA", "DISHA") and display all rows where the base name repeats. Use string manipulation to extract the base name."""
#-----------------------------Write Solution Code-----------------------------

df = pd.read_excel("CSEB_Attendance.xlsx", sheet_name="Student Summary")
df["Base Name"] = df["Student Name"].str.split().str[0].str.upper()
duplicates = df[df.duplicated(subset="Base Name", keep=False)]
print("Duplicate Student Names:")
print(duplicates)
```

```python
""""Q19. Merge this dataset with a hypothetical Student_Details DataFrame containing Roll No, Branch, and Semester. Perform an inner join and display the merged result."""
#-----------------------------Write Solution Code-----------------------------

df = pd.read_excel("CSEB_Attendance.xlsx", sheet_name="Student Summary")
Student_Details = pd.DataFrame({
    "Roll No": ["4MW24CS061", "4MW24CS062", "4MW24CS063", "4MW24CS064"],
    "Branch": ["CSE", "CSE", "CSE", "CSE"],
    "Semester": [5, 5, 5, 5]
})
merged_df = pd.merge(
    df,
    Student_Details,
    on="Roll No",
    how="inner"
)
print("Merged Student Details:")
print(merged_df)

```

```python
"""Q20. Export the following to separate CSV files:
(a) All "Critical" students
(b) Summary statistics grouped by Status
(c) Top 10 students by Risk Score"""
#-----------------------------Write Solution Code-----------------------------

df = pd.read_excel("CSEB_Attendance.xlsx", sheet_name="Student Summary")
df["Attendance %"] = (
    df["Attendance %"].astype(str).str.replace("%", "").astype(float)
)
critical_students = df[df["Status"] == "Critical"]
critical_students.to_csv("Critical_Students.csv", index=False)
summary = df.groupby("Status").agg({
    "Present": ["min", "max", "mean"],
    "Absent": "sum",
    "Student Name": "count"
})
summary.to_csv("Status_Summary.csv")
df["Risk Score"] = (
    df["Absent"] * 0.4 +
    df["Consecutive Absences"] * 0.6
)
top_10 = df.sort_values(
    by="Risk Score", ascending=False
).head(10)
top_10.to_csv("Top_10_Risk_Score.csv", index=False)
print("All three CSV files have been exported successfully!")
```
